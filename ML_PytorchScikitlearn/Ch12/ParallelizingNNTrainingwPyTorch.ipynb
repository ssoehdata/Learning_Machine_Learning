### Applying Mathematical Operaitons to Tensors

```python
torch.manual_seed(1)
t1 = 2 * torch.rand(5, 2) -1 
t2 = torch.normal(mean=0, std=1, size=(5, 2))
```

```python
t3 = torch.multiply(t1, t2)
print(t3)
```

```python
t4 = torch.mean(t1, axis=0)
print(t4)
```

```python
t5 = torch.matmul(t1, torch.transpose(t2, 0, 1))
print(t5)
```

```python
t6 = torch.matmul(torch.transpose(t1, 0, 1), t2)
print(t6)
```

```python
norm_t1 = torch.linalg.norm(t1, ord=2, dim=1)
print(norm_t1)
```

### Split, Stack, and Concatenate Tensors

```python
torch.manual_seed(1)
t = torch.rand(6)
print(t)
```

```python
t_splits = torch.chunk(t, 3)
[item.numpy() for item in t_splits]
```

#### Specify Sizes of Output Tensors Directly (instead of defining the number of splits)

```python
torch.manual_seed(1)
t = torch.rand(5)
print(t)
```

```python
t_splits = torch.split(t, split_size_or_sections=[3,2])
[item.numpy() for item in t_splits]
```

### Concatenate Multiple Tensors

```python
A = torch.ones(3)
B = torch.zeros(2)
C = torch.cat([A, B], axis=0)
print(C)
```

### Stacking 1D Tensors to Form A 2D Tensor

```python
A = torch.ones(3)
B = torch.zeros(3)
S = torch.stack([A, B], axis=1)
print(S)
```

### Combining Two Tensors Into A Joint Dataset

```python
torch.manual_seed(1)
t_x = torch.rand([4, 3], dtype=torch.float32)
t_y = torch.arange(4)
```

```python
from torch.utils.data import TensorDataset
joint_dataset = TensorDataset(t_x, t_y)
```

```python
for example in joint_dataset:
    print(' x: ', example[0], ' y: ', example[1])
```

```python
joint_dataset = TensorDataset(t_x, t_y)
```

```python
torch.manual_seed(1)
data_loader = DataLoader(dataset=joint_dataset, batch_size=2, shuffle=True)
```

```python
for i, batch in enumerate(data_loader, 1):
    print(f'batch {i}:', 'x:', batch[0],
          '\n             y:', batch[1])
```

### Iterate over the Batched Dataset Twice:

```python
for epoch in range(2):
    print(f'Epoch {epoch+1}')
    for i, batch in enumerate(data_loader, 1):
        print(f'Batch {i}:', 'x:', batch[0],
              '\n             y:', batch[1])
```
